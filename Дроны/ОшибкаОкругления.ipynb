```python
import struct
import time

def float32_to_exact_value(f):
    """Преобразует float64 (Python) в float32 (как в инфузомате) и обратно в точное десятичное приближение"""
    # Упаковываем как 32-битное число (как в инфузомате)
    packed = struct.pack('f', f)      # 'f' = float32
    # Распаковываем обратно — теперь это значение с ошибкой округления
    rounded = struct.unpack('f', packed)[0]
    return rounded

def main():
    print("💉 СИМУЛЯТОР ИНФУЗОМАТА: ОШИБКА ОКРУГЛЕНИЯ В ДЕЙСТВИИ")
    print("Многие дробные числа НЕЛЬЗЯ точно представить в двоичном виде.")
    print("Это может привести к небольшой, но критичной ошибке в дозе!\n")
    time.sleep(1)

    try:
        # Ввод дозы
        dose_input = input("Введите дозу лекарства (мкг/кг/мин): ")
        dose_target = float(dose_input)

        if dose_target <= 0:
            print("❌ Доза должна быть положительной.")
            return

        # Эмуляция хранения в инфузомате (32-битный float)
        dose_actual = float32_to_exact_value(dose_target)

        # Расчёт ошибки
        absolute_error = abs(dose_actual - dose_target)
        relative_error_percent = (absolute_error / dose_target) * 100 if dose_target != 0 else 0

        # Накопление за 1 час (60 минут)
        total_target = dose_target * 60
        total_actual = dose_actual * 60
        total_error = abs(total_actual - total_target)

        # Вывод результатов
        print("\n" + "="*60)
        print("📊 РЕЗУЛЬТАТЫ СИМУЛЯЦИИ")
        print("="*60)
        print(f"Введённая доза:       {dose_target:.6f} мкг/кг/мин")
        print(f"Фактическая доза:     {dose_actual:.16f} мкг/кг/мин")
        print(f"Абсолютная ошибка:    {absolute_error:.16f} мкг/кг/мин")
        print(f"Относительная ошибка: {relative_error_percent:.4f}%")
        print()
        print(f"За 1 час пациент получит:")
        print(f"  По назначению:      {total_target:.3f} мкг/кг")
        print(f"  Фактически:         {total_actual:.13f} мкг/кг")
        print(f"  Разница:            {total_error:.13f} мкг/кг")
        print("="*60)

        # Предупреждение, если ошибка значима
        if relative_error_percent > 0.1:
            print("\n⚠️  ВНИМАНИЕ: ошибка превышает 0.1%!")
            print("   Для катехоламинов (допамин, норадреналин) это может быть клинически значимо.")
        elif absolute_error > 0.01:
            print("\nℹ️  Небольшая ошибка. Обычно безопасна, но требует контроля.")
        else:
            print("\n✅ Ошибка минимальна. В пределах допуска.")

        print("\n💡 Вывод: всегда сверяйте дозу визуально и при сомнениях — перепроверяйте!")

    except ValueError:
        print("❌ Ошибка: введите корректное число (например, 5.1 или 10).")
    except KeyboardInterrupt:
        print("\n\n👋 Симуляция завершена.")

if __name__ == "__main__":
    main()
```
